```python
import re

def chemspace_label_to_smarts(label : str) -> str:
    '''req import re'''
    # Chemspace uses [R]'s pretty much synonymously to wildcard atoms
    label = re.sub(r'\[R\]', '*', label)

    def repl(match):
        # Chemspace labels feature 'ar' (aromatic) and 'al' (alifatic) right after atom letters to convey aromaticity
        # removing and lowering/capitalizing the previous letter makes it smarts friendly without info loss
        prev_char = match.group(1)
        pattern = match.group(2)
        if pattern == 'ar':
            return prev_char.lower()  # lowercase for 'ar'
        elif pattern == 'al':
            return prev_char.upper()  # uppercase for 'al'

    # (.) captures the previous character, (ar|al) captures the pattern
    return re.sub(r'(.)((?:ar|al))', repl, label)
```

```python
chemspace_structs = pd.read_csv('chemspaceweb_2023_structures.csv')[['label','smarts']].dropna()
```

```python
chemspace_structs["inferred_smarts"] = chemspace_structs["label"].apply(chemspace_label_to_smarts)
chemspace_structs["inferred_smarts"] = chemspace_structs["inferred_smarts"].apply(lambda x : Chem.MolToSmarts(Chem.MolFromSmarts(x, mergeHs=True)))
```

```python
df = pd.read_csv(data_dir + '/chemspace_checkpoint.csv')
```

```python
cids_props = pd.read_csv(data_dir + '/CIDs_to_mass_iupac.csv')
```

```python
all_cids = set([int(cid) for cids_list in df['cids'] for cid in eval(cids_list)])
```

```python
cids_props = cids_props[cids_props['CID'].isin(all_cids)].copy()
```

```python
# Prepare top N smallest MW IUPAC names per SMARTS
best_IUPAC_lists = []
import ast

for i, row in df.iterrows():
    cids_list = row['cids']
    # safely convert stringified lists
    if isinstance(cids_list, str):
        try:
            cids_list = ast.literal_eval(cids_list)
        except Exception:
            cids_list = []
            

    if not cids_list:
        best_IUPAC_lists.append([])
        continue

    subset = cids_props[cids_props['CID'].isin(cids_list)]
    subset_sorted = subset.sort_values('Mass', ascending=True)

    # Get top_n IUPAC names
    top_names = subset_sorted['IUPAC'].head(10).to_list()
    best_IUPAC_lists.append(top_names)

df['best_IUPAC_names'] = best_IUPAC_lists
```

```python
few_shot_text = "Here are IUPAC names of some of the smaller compounds this smarts pattern matches, use them as inpiration for naming the smarts pattern, but keep in mind that the smarts pattern might be a lot simpler than even the simplest compound it matches, so dont just automatically take the name of the smallest compound:\n"
prompts = []
for iupac_name_list in df['best_IUPAC_names']:
    if iupac_name_list == []:
        prompts.append('')
    else:
        prompts.append(
            few_shot_text + "\n".join(iupac_name_list)
        )
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem, DataStructs
from rdkit.Chem.rdFingerprintGenerator import GetMorganGenerator
from tqdm import tqdm 
def get_top_n_most_similar_smarts_description_examples(query_fp, ref_df, n):
    # Compute similarity for each fingerprint
    if query_fp == None:
        return list(ref_df.sample(2 * n)
                          [['smarts', 'cleaned_description']]
                          .itertuples(index=False, name=None)), random
    similarities = []
    for idx, row in ref_df.iterrows():
        fp2 = row['fps']
        if fp2 is not None:
            sim = DataStructs.TanimotoSimilarity(query_fp, fp2)
            similarities.append(((row['smarts'], row['cleaned_description']), sim))  # store SMARTS with similarity
    
    # Sort by similarity descending
    similarities.sort(key=lambda x: x[1], reverse=True)
    
    # Get top n SMARTS-description pairs
    return [s for s, sim in similarities[:n]], random
```

```python
def get_top_n_most_similar_smarts_description_examples_wrapper(df, smarts_col, n=10):
    ref = pd.read_csv(data_dir + '/desc_refs.csv')
    
    # Initialize Morgan fingerprint generator
    morgan_gen = GetMorganGenerator(radius=2, fpSize=1024)
    
    def get_fps_from_list(smarts_list):
        fps = []
        for s in smarts_list:
            mol = Chem.MolFromSmarts(s)
            
            if mol is not None:
                try:
                    Chem.SanitizeMol(mol)
                    fp = morgan_gen.GetFingerprint(mol)
                    fps.append(fp)
                except Exception as e:
                    fps.append(None)
            else:
                fps.append(None)
                print(f"Invalid SMARTS: {s}")
        return fps

    ref['fps'] = get_fps_from_list(ref['smarts'].to_list())
    ref.dropna(inplace=True)
    smarts_fps = get_fps_from_list(df[smarts_col].to_list())

    smarts_examples = []
    for query_smarts in tqdm(smarts_fps, desc="Finding most similar smarts examples"):
        examples, are_random = get_top_n_most_similar_smarts_description_examples(query_smarts, ref, n=n)
        smarts_examples.append(examples)

    df['similar_examples'] = smarts_examples

    return df
    
df = get_top_n_most_similar_smarts_description_examples_wrapper(df, 'inferred_smarts', n=10)
```

```python
shared_base_instructions = """You are an LLM that converts SMARTS substructure filters into concise, human-readable names. 
- **Input**: A SMARTS string. 
- **Output**: A lowercase name with words separated by spaces. 
- The name should help molecular biologists/chemists quickly grasp the filter's function (e.g., 'amide bond'). 
- Return **only the name**, no prefixes/suffixes/explanations.
"""
matching_compounds_example_prompt_base = "Here are IUPAC names of some of the smaller compounds this smarts pattern matches, use them as inpiration for naming the smarts pattern, but keep in mind that the smarts pattern might be a lot simpler than even the simplest compound it matches, so dont just automatically take the name of the smallest compound:\n"
similar_examples_prompt_base = "\nHere are examples of filters and their corresponding name, use them as inspiration for how filters tend to be named in relation to their smarts pattern:\n"
def build_prompt(best_IUPAC_names, similar_examples, shared_base_instructions, input_smarts):
    matching_compounds_example_prompt = matching_compounds_example_prompt_base + '\n'.join(best_IUPAC_names) if best_IUPAC_names != [] else ''
    similar_examples_prompt = similar_examples_prompt_base + ''.join([f"Input: {smarts} -> Output: {name}\n" for smarts, name in examples])
    return shared_base_instructions + matching_compounds_example_prompt + similar_examples_prompt + f"Input: {input_smarts} -> Output:"
```

```python
i = 1
print(build_prompt(df.iloc[i]['best_IUPAC_names'], df.iloc[i]['similar_examples'], shared_base_instructions, df.iloc[i]['inferred_smarts']))
```
